# Visualising 

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("data/Daily_Responses_5_Years_2019_full.csv")
data.head()
```

```python
# Filtering to Trust Area
Trust = data[data["area"] == "Trust"]
plt.figure(figsize=(13, 5))
plt.plot(Trust["ds"], Trust["y"])
plt.xlabel("Date")
plt.ylabel("Daily Responses")
plt.title("Number of Daily Responses in Trust between 2013-2019 ")
plt.xticks(Trust["ds"][::365])
plt.xticks(rotation=45)
plt.show()
```

```python
# repating the code above for each area - making a dictionary for each area:
areas = ["Cornwall", "Gloucestershire", "Wiltshire", "Dorset", "BNSSG", "Somerset"]

for area_name in areas:
    area_data = data[data["area"] == area_name]
    plt.figure(figsize=(13, 5))
    plt.plot(area_data["ds"], area_data["y"])
    plt.xlabel("Date")
    plt.ylabel("Daily Responses")
    plt.title("Number of Daily Responses in " + area_name + " between 2013-2019")
    plt.xticks(area_data["ds"][::365])
    plt.xticks(rotation=45)
    plt.show()
```
